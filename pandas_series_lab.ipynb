```python
#Marks Analysis with Conditions
import pandas as pd

# Dataset
marks = [45, 67, 89, 34, 90, 55, 72]
subjects = ['Math', 'Physics', 'Chemistry', 'English', 'CS', 'Biology', 'History']

# Create Pandas Series
s = pd.Series(marks, index=subjects)

# 1️⃣ Subjects with marks > 70
print("Subjects with marks > 70:")
print(s[s > 70])

# 2️⃣ Replace marks < 40 with 'Fail'
s_modified = s.copy()
s_modified[s_modified < 40] = "Fail"
print("\nMarks after replacing <40 with 'Fail':")
print(s_modified)

# 3️⃣ Count how many subjects scored above average
average_marks = s.mean()
count_above_avg = (s > average_marks).sum()
print(f"\nNumber of subjects above average ({average_marks}): {count_above_avg}")
```

```python
#Salary Increment Logic
# Dataset
salary = [25000, 32000, 28000, 40000, 50000]
emp = ['A', 'B', 'C', 'D', 'E']

# Create Series
s = pd.Series(salary, index=emp)

# 1️⃣ Increase salary by 10% where salary < 30000
s[s < 30000] *= 1.10

# 2️⃣ Decrease salary by 5% where salary > 45000
s[s > 45000] *= 0.95

# 3️⃣ Total salary expense
total_salary = s.sum()

print("Updated Salaries:")
print(s)
print(f"\nTotal Salary Expense: {total_salary}")
```

```python
#Handling Missing Data Smartly
# Dataset
data = [10, None, 30, None, 50, 60, None]
s = pd.Series(data)

# Count missing values before filling
missing_before = s.isna().sum()

# Fill missing values with mean
s.fillna(s.mean(), inplace=True)

# Count missing values after filling
missing_after = s.isna().sum()

# Drop remaining nulls (if any)
s.dropna(inplace=True)

print("Series after filling missing values:")
print(s)
print(f"\nMissing before: {missing_before}, Missing after: {missing_after}")
```

```python
#Student Grades System
# Dataset
marks = [95, 82, 67, 58, 73, 49, 88]
students = ['S1','S2','S3','S4','S5','S6','S7']

# Create Series
s = pd.Series(marks, index=students)

# Function to assign grades
def grade(mark):
    if mark >= 90:
        return 'A'
    elif mark >= 75:
        return 'B'
    elif mark >= 60:
        return 'C'
    else:
        return 'D'

# Assign grades using apply()
grades = s.apply(grade)

# Count number of students in each grade
grade_counts = grades.value_counts()

print("Grades assigned:")
print(grades)
print("\nGrade counts:")
print(grade_counts)
```

```python
#Series Alignment
# Dataset
s1 = pd.Series([10, 20, 30], index=['A', 'B', 'C'])
s2 = pd.Series([5, 15, 25], index=['B', 'C', 'D'])

# Add both series (NaN for non-matching indexes)
s_sum = s1 + s2
print("Sum with NaN:")
print(s_sum)

# Replace NaN with 0 and recompute
s_sum_filled = (s1 + s2).fillna(0)
print("\nSum with NaN replaced by 0:")
print(s_sum_filled)
```

```python
#Ranking System
# Dataset
scores = [88, 92, 79, 93, 85]
players = ['P1','P2','P3','P4','P5']

# Create Series
s = pd.Series(scores, index=players)

# Rank players (highest score = rank 1)
rank = s.rank(ascending=False, method='min')

# Top 3 players
top3 = s.nlargest(3)

# Player with lowest score
lowest = s.idxmin()

print("Player Rankings:")
print(rank)
print("\nTop 3 players:")
print(top3)
print(f"\nPlayer with lowest score: {lowest}")
```

```python
#String Operations in Series
# Dataset
names = ['anuj', 'rahul', 'sneha', 'kiran', 'amit']
s = pd.Series(names)

# Convert all names to uppercase
upper_names = s.str.upper()

# Find names containing letter 'a'
contains_a = s[s.str.contains('a')]

# Replace 'anuj' with 'Anuj Kumar'
s_replaced = s.replace('anuj', 'Anuj Kumar')

print("Uppercase names:")
print(upper_names)
print("\nNames containing 'a':")
print(contains_a)
print("\nNames after replacement:")
print(s_replaced)
```

```python
#Chained Operations
# Dataset
data = [5, None, 15, 25, None, 35, 45]
s = pd.Series(data)

# Fill missing with median -> filter values > 20 -> sort descending -> reset index
result = (s.fillna(s.median())
            .loc[lambda x: x > 20]
            .sort_values(ascending=False)
            .reset_index(drop=True))

print("Result after chained operations:")
print(result)
```

```python
#Sales Performance
import matplotlib.pyplot as plt

# Dataset
sales = [200, 450, 300, 150, 500, 700, 100]
days = ['Mon','Tue','Wed','Thu','Fri','Sat','Sun']

# Create Series
s = pd.Series(sales, index=days)

# Best and worst sales day
best_day = s.idxmax()
worst_day = s.idxmin()

# Total and average sales
total_sales = s.sum()
average_sales = s.mean()

# Apply 20% bonus where sales > 400
s[s > 400] *= 1.20

# Sort sales descending
s_sorted = s.sort_values(ascending=False)

print(f"Best day: {best_day}, Worst day: {worst_day}")
print(f"Total sales: {total_sales}, Average sales: {average_sales}")
print("\nSales after bonus and sorted:")
print(s_sorted)

# Optional: Plot sales
s_sorted.plot(kind='bar', title="Sales after Bonus", color='skyblue')
plt.show()
```
