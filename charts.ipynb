```python
# Step 1: Import Matplotlib
import matplotlib.pyplot as plt

# Step 2: Data
categories = ['Apples', 'Bananas', 'Cherries', 'Dates']
values = [10, 15, 7, 12]

# Step 3: Create bar chart
plt.figure(figsize=(8,5))                # Figure size
plt.bar(categories, values, color='skyblue')  # Bar chart with color
plt.title("Fruit Count")                 # Chart title
plt.xlabel("Fruits")                     # X-axis label
plt.ylabel("Quantity")                   # Y-axis label
plt.grid(axis='y', linestyle='--', alpha=0.7) # Y-axis grid
plt.show()                               # Display chart
```

```python
# Step 1: Import Matplotlib
import matplotlib.pyplot as plt

# Step 2: Data
products = ['Laptop', 'Phone', 'Tablet', 'Monitor', 'Keyboard', 'Mouse']
sales = [120, 250, 90, 70, 150, 200]
colors = ['red', 'blue', 'green', 'orange', 'purple', 'cyan']  # Different color for each bar

# Step 3: Create bar chart
plt.figure(figsize=(10,6))                           # Set figure size
plt.bar(products, sales, color=colors)               # Bar chart with different colors
plt.title("Sales of Products")                       # Chart title
plt.xlabel("Products")                               # X-axis label
plt.ylabel("Units Sold")                             # Y-axis label
plt.grid(axis='y', linestyle='--', alpha=0.7)       # Add Y-axis grid lines
plt.show()                                           # Display chart
```

```python
# Step 1: Import Matplotlib
import matplotlib.pyplot as plt

# Step 2: Data
days = list(range(1, 31))  # Days from 1 to 30
temperature = [22, 24, 23, 25, 26, 24, 23, 22, 21, 24, 
               25, 27, 26, 24, 23, 22, 21, 23, 24, 25,
               26, 27, 28, 26, 25, 24, 23, 22, 21, 23]  # Example temperatures

# Step 3: Create line chart
plt.figure(figsize=(12,6))
plt.plot(days, temperature, marker='o', color='orange', linestyle='-', linewidth=2)
plt.title("Daily Temperature for 30 Days")   # Chart title
plt.xlabel("Day")                             # X-axis label
plt.ylabel("Temperature (°C)")               # Y-axis label
plt.grid(True, linestyle='--', alpha=0.7)    # Grid lines
plt.xticks(days)                              # Show all days on X-axis
plt.show()
```

```python
# Step 1: Import Matplotlib
import matplotlib.pyplot as plt

# Step 2: Data
place = "Delhi"
days = list(range(1, 31))  # Days from 1 to 30
temperature = [28, 30, 29, 31, 32, 30, 29, 28, 27, 30, 
               31, 33, 32, 30, 29, 28, 27, 29, 30, 31,
               32, 33, 34, 32, 31, 30, 29, 28, 27, 29]  # Example temperatures

# Step 3: Create line chart
plt.figure(figsize=(12,6))
plt.plot(days, temperature, marker='o', color='red', linestyle='-', linewidth=2)
plt.title(f"Daily Temperature for 30 Days in {place}")   # Title includes place name
plt.xlabel("Day")                                         # X-axis label
plt.ylabel("Temperature (°C)")                            # Y-axis label
plt.grid(True, linestyle='--', alpha=0.7)                # Grid lines
plt.xticks(days)                                         # Show all days on X-axis
plt.show()
```
